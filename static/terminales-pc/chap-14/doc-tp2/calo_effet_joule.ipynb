```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('ggplot')
```

# Élévation de la température d'une masse $m$ d'eau à l'aide d'un courant électrique.


**Z1** : Valeurs expérimentales

```python
U = 5.93   # V
I = 1.53   # A

C = 104   # J/°C/g
c = 4.18   # J/°C/g
m = 140    # g

dates = np.linspace(0, 420, 15)     # Liste des dates en seconde
thetas = np.linspace(23.2, 29, 15)  # Liste des températures
```

**Z2** : Calcul de la puissance $P$.

```python
P = U * I
```

**Z3** : Détermination des valeurs de $Q$ en fonction du temps.

```python
Q = []

for t in dates:
    Q.append(P * t)
```

**Z4** : Détermination des valeurs de $\Delta U$ en fonction du temps.

```python
delta_U = []
tetha_0 = thetas[0]

for theta in thetas:
    delta_U.append( (C + m * c) * (theta - theta_0) )
```

**Z5** : Tracés de $Q = f(t)$ et $\Delta U = g(t)$. 

```python
plt.figure(figsize=(8, 6), dpi=100)
plt.plot( dates , Q , 'o', label="$Q$ (J)")
plt.plot( dates , delta_U , 'o', label="$\Delta U$ (J)")
plt.xlabel("t (s)")
plt.legend()
plt.show() 
```

```python
print(f"Énergie thermique transférée au système : {Q[-1]} (J)")
print(f"Variation de l'énergie interne du système : {delta_U[-1]} (J)")
print(f"Erreur relative : {abs(Q[-1] - delta_U[-1]) / Q[-1]}")
```
